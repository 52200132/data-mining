## Tạo vector đặc trưng sử dụng TF IDF

```python
import re


def load_data(file_path) -> list[str]:
    data = []
    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            if (line.strip() == ""):
                continue
            clean_text = re.sub(r"\d+", "", line.strip())
            data.append(clean_text)
    return data
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Dữ liệu mẫu (Tập văn bản)
my_datas = load_data("./processed-data/data_1776780928.txt")


vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(my_datas)
```

```python
words = vectorizer.get_feature_names_out()
print("Từ điển:", words)

# 5. Xem ma trận TF-IDF dưới dạng mảng (Array)
print("Ma trận TF-IDF:\n", tfidf_matrix.toarray())
```

## Thực hiện gom cụm dữ liệu bằng các thuật toán đã học như: K-Mean, Hierarchical, DBSCAN, số cụm là 5

```python
# 1. Gom cụm với K-Means
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
import pandas as pd

data_vectorizered = tfidf_matrix.toarray()

kmeans = KMeans(n_clusters=5, random_state=42)
kmeans_labels = kmeans.fit_predict(data_vectorizered)

# 2. Gom cụm với Hierarchical
hierarchical = AgglomerativeClustering(n_clusters=5)
hierarchical_labels = hierarchical.fit_predict(data_vectorizered)

# 3. Gom cụm với DBSCAN (Cần tinh chỉnh eps cho phù hợp với dữ liệu TF-IDF)
dbscan = DBSCAN(eps=1.5, min_samples=2)
dbscan_labels = dbscan.fit_predict(data_vectorizered)

# Lưu kết quả vào file CSV
df = pd.DataFrame(
    {
        "Văn bản": my_datas,
        "K-Means": kmeans_labels,
        "Hierarchical": hierarchical_labels,
        "DBSCAN": dbscan_labels,
    }
)

df.to_csv("clustering_results.csv", index=False, encoding="utf-8-sig")
```

```python
# 3. Gom cụm với DBSCAN (Cần tinh chỉnh eps cho phù hợp với dữ liệu TF-IDF)
dbscan = DBSCAN(eps=1.3, min_samples=2)
dbscan_labels = dbscan.fit_predict(data_vectorizered)

print("Kết quả gom cụm với DBSCAN:", dbscan_labels)
```
